Python: String Manipulation

A string is a sequence of characters enclosed in quotes.

```python
name = "Maham"
message = 'Hello, Python!'
```

```python
#1. Common String Methods
#upper() — Convert to uppercase

name = "maham"
print(name.upper())
```

```python
#lower() — Convert to lowercase

name = "MAHAM"
print(name.lower())
```

```python
#capitalize() — Capitalize the first letter
text = "python is easy"
print(text.capitalize())
```

```python
#title() — Capitalize the first letter of each word
text = "python programming"
print(text.title())
```

```python
#strip() — Remove spaces from beginning and end
text = "  Hello  "
print(text.strip())
```

```python
#replace() — Replace part of a string
text = "I like Java"
print(text.replace("Java", "Python"))
```

```python
#split() — Split a string into a list
text = "Python is easy"
words = text.split()

print(words)
```

```python
#join() — Join elements together
words = ["Python", "is", "easy"]

sentence = " ".join(words)
print(sentence)
```

```python
#find() — Find the position of a substring
text = "I love Python"

print(text.find("Python"))
```

```python
#count() — Count occurrences
text = "banana"

print(text.count("a"))
```

2. F-String Formatting

f-strings allow us to easily insert variables inside a string.

We write f before the quotation marks and put variables inside {}.

```python
name = "Maham"
age = 24

print(f"My name is {name} and I am {age} years old.")
```

```python
#Without f-string
name = "Maham"
age = 24

print("My name is", name, "and I am", age, "years old.")
```

```python
#With f-string
print(f"My name is {name} and I am {age} years old.")
```

```python
#3. Expressions Inside F-Strings

#You can perform calculations inside {}.
a = 10
b = 20

print(f"The sum is {a + b}")
```

```python
#4. Formatting Decimal Numbers

#You can control the number of decimal places using :.2f.
price = 25.6789

print(f"The price is ${price:.2f}")
```

```python
#5. Example: Student Information
name = "Maham"
subject = "Molecular Biology"
cgpa = 3.92

print(f"Student: {name}")
print(f"Subject: {subject}")
print(f"Marks: {cgpa}")
```
