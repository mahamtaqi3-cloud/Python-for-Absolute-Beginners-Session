Polymorphism and encapsulation are two of the main principles of Object-Oriented Programming (OOP), along with inheritance.

1. Polymorphism

Poly = many
Morph = forms

It allows different classes to use the same method name but perform different actions.

Think of the word "speak()":

A dog speaks by barking.
A cat speaks by meowing.
A bird speaks by chirping.

The method name is the same, but each object performs it differently.



```python
#Example 1: Animal Sounds
class Dog:
    def speak(self):
        print("Woof!")

class Cat:
    def speak(self):
        print("Meow!")

class Bird:
    def speak(self):
        print("Chirp!")

dog = Dog()
cat = Cat()
bird = Bird()

dog.speak()
cat.speak()
bird.speak()
```

Notice:

The method is always called
speak()

but the result changes according to the object.

This is polymorphism.

```python
#Example 2: Using a Loop

#One of the biggest advantages of polymorphism is that we don't need to care which object we're using.
class Dog:
    def speak(self):
        print("Woof!")

class Cat:
    def speak(self):
        print("Meow!")

class Cow:
    def speak(self):
        print("Moo!")

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.speak()
```

```python
#The loop simply calls
animal.speak()
#Python figures out which version to use.
```

2. Encapsulation

Encapsulation means bundling data (attributes) and methods (functions) together in a class and restricting direct access to some of the object's data.


Think of it as a capsule.

A medicine capsule hides what's inside. You use it through the outside interface.

Similarly, a class can hide its internal data and expose controlled methods.

```python
#Without Encapsulation
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s = Student("Ali", 85)

print(s.name)
print(s.marks)

s.marks = -100

print(s.marks)
```

```python
#Using Encapsulation

#Python uses naming conventions to indicate that an attribute is intended to be private.
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks

    def show_marks(self):
        print(self.__marks)

print(s.name)

s = Student("Ali", 85)

s.show_marks()
```

The variable

__marks

is name-mangled by Python to discourage direct access.

Trying

print(s.__marks)

gives

AttributeError

because Python internally stores it with a different name (for example, _Student__marks).

```python
print(s.__marks)
```
