1. Student Class (Beginner ⭐)

Create a class called Student.

Attributes:

name
age
grade

Methods:

display_info() → prints all student details.
study() → prints:

Ali is studying.

Practice:

Create two student objects.
Display their information.
Call the study() method.

```python
#Code Solution
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def display_info(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Grade:", self.grade)

    def study(self):
        print(f"{self.name} is studying.")


student1 = Student("Ali", 20, "A")
student2 = Student("Sara", 19, "B")

student1.display_info()
student2.display_info()

student1.study()
student2.study()



```

2. BankAccount Class ⭐

Create a class named BankAccount.

Attributes:

account_holder
balance

Methods:

deposit(amount)
withdraw(amount)
show_balance()

Rules:

Don't allow withdrawing more than the available balance.
Print an appropriate message if there isn't enough money.

Example:

Current Balance: $1500
Deposited: $500
Current Balance: $2000

```python
#2 Solution
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"${amount} deposited.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"${amount} withdrawn.")

        else:
            print("Insufficient balance!")

    def show_balance(self):
        print("Current Balance: $", self.balance)

account = BankAccount("Ali", 1000)

account.show_balance()
account.deposit(500)
account.withdraw(200)
account.show_balance()


```

3. Rectangle Class ⭐

Create a class named Rectangle.

Attributes:

length
width

Methods:

area()
perimeter()

Formula:

Area = length × width
Perimeter = 2 × (length + width)

```python
#Rectangle class
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r = Rectangle(10, 5)
print("Area:", r.area())
print("Perimeter:", r.perimeter())


```

4. Car Class ⭐⭐

Create a class named Car.

Attributes:

brand
model
year

Methods:

start_engine()
stop_engine()
display_info()

Example output:

Toyota Corolla (2022)
Engine Started.

```python
#Car Class
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def start_engine(self):
        print("Engine Started.")

    def stop_engine(self):
        print("Engine Stopped.")

    def display_info(self):
        print(self.brand, self.model, self.year)

car = Car("Toyota", "Corolla", 2025)
car.display_info()
car.start_engine()


```

```python
#Book Class
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def read(self):
      print(f"You are reading '{self.title}'.")

    def book_info(self):
        print("Title:", self.title)
        print("Author:", self.author)
        print("Pages:", self.pages)

book1 = Book("Python basics", "John", 250)
book2 = Book("Data science", "Alice", 300)
book3 = Book("AI", "David", 450)

book1.book_info()
book1.read()

book2.book_info()
book2.read()

book3.book_info()
book3.read()
```
