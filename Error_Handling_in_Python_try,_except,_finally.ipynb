Error Handling in Python: try, except, finally

When Python encounters an error, your program normally stops running. Error handling allows us to handle these errors gracefully instead of letting the program crash unexpectedly.

The main keywords are:

try → Code that might cause an error

except → What to do if an error occurs

finally → Code that always runs, whether there is an error or not

```python
#1. Basic try and except
try:
    number = int(input("Enter a number: "))
    print(number)
except:
    print("Something went wrong!")
```

2. Handling a Specific Error

It is better to specify the type of error you expect.

```python
try:
    number = int(input("Enter a number: "))
    print(number)
except ValueError:
    print("Please enter a valid number!")
```

Here, ValueError occurs when Python cannot convert the input into an integer.

```python
#3. Multiple except Blocks

#You can handle different errors differently.
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    result = number1 / number2
    print(result)

except ValueError:
    print("Please enter numbers only!")

except ZeroDivisionError:
    print("You cannot divide by zero!")
```

```python
4#. finally

#The finally block always executes.
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Invalid input!")

finally:
    print("Program finished.")
```

try      → Try to run this code

except   → Handle the error

finally  → Always run this code

#Easy way to remember

try:
    # Try something

except:
    # If it fails, handle the error
    
finally:
    # Always do this

```python
#Real-world example

#Imagine you're reading a file:
try:
    file = open("data.txt", "r")
    data = file.read()
    print(data)

except FileNotFoundError:
    print("The file does not exist.")

finally:
    print("Finished trying to read the file.")
```

This is useful because your program doesn't immediately stop when the file is missing.

⚠️ Important tip

Avoid using a completely empty except like this:

```python
try:
    # code
except:
    # something
```

```python
#Prefer specific errors:
except ValueError:
```

```python
except FileNotFoundError:
```

This makes your code easier to debug and prevents you from accidentally hiding unexpected errors.

In short:

try = "Try this code."

except = "If something goes wrong, handle it."

finally = "No matter what happens, do this."
