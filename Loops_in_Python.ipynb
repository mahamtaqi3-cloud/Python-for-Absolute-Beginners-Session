In Python, a loop is used to repeat a block of code multiple times.

There are two main types of loops:

1. for Loop

A for loop is used when you want to repeat something for each item in a sequence, such as a list or a range of numbers.

Syntax: for variable in sequence:
    # code to repeat

````python
#Example
for i in range(5):
    print(i)
````

Here:

for → starts the loop

i → loop variable

range(5) → generates numbers from 0 to 4

print(i) → runs repeatedly

````python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
````

````python
for i in range(5):
    print("Hello")
````

````python
for i in range(1, 6):
    print(i)
````

2. Loop Through a List

Suppose we have a list of fruits:

````python
fruits = ["Apple", "Banana", "Mango", "Orange"]

for fruit in fruits:
    print(fruit)
````

3. Loop Through a String

A string is also a sequence of characters.

````python
name = "Maham"

for letter in name:
    print(letter)
````

4. for Loop with Calculation

You can perform calculations inside a loop.

Example: Print squares

````python
for i in range(1, 6):
    print(i * i)
````

2. while Loop

A while loop repeats code as long as a condition is True.

Syntax:

while condition:

    # code to repeat

````python
i = 1

while i <= 5:
    i = i + 1
    print(i)
````

````python
i = 1

while i <= 5:
  print(i)
````

How it works:

Initially:

count = 1

Python checks:

count <= 5

If it is True, the code runs.

Then:

count = count + 1

increases the value.

The process continues:

count = 1 → print
count = 2 → print
count = 3 → print
count = 4 → print
count = 5 → print
count = 6 → stop


````python
#continue

#continue is used to skip the current iteration and move to the next one.

for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````

````python
names = ["Ali", "Sara", "Ahmed"]

for name in names:
    if name == "Sara":
        continue
    print(name)
````

Loop with if Condition

You can combine loops and conditions.

Example: Find even numbers



````python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
````

## 🔄 What is a Nested Loop?

A **nested loop** simply means:

> **One loop inside another loop.**

Think of it like this:

```text
Outer Loop
   ↓
   Inner Loop
      ↓
      Code runs
```

The **inner loop completes all its repetitions** every time the **outer loop runs once**.

---

## 🪜 Simple Example

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

Let's break it down.

### Outer loop:

```python
for i in range(1, 4):
```

The outer loop gives:

```text
i = 1
i = 2
i = 3
```

### Inner loop:

```python
for j in range(1, 4):
```

The inner loop also gives:

```text
j = 1
j = 2
j = 3
```

Now let's see what happens.

---

### 🟢 First: `i = 1`

The inner loop runs completely:

```text
i = 1, j = 1 → print 1 1
i = 1, j = 2 → print 1 2
i = 1, j = 3 → print 1 3
```

Output:

```text
1 1
1 2
1 3
```

---

### 🔵 Second: `i = 2`

Again, the **entire inner loop runs**:

```text
i = 2, j = 1 → print 2 1
i = 2, j = 2 → print 2 2
i = 2, j = 3 → print 2 3
```

Output:

```text
2 1
2 2
2 3
```

---

### 🟠 Third: `i = 3`

Again, the inner loop runs completely:

```text
i = 3, j = 1 → print 3 1
i = 3, j = 2 → print 3 2
i = 3, j = 3 → print 3 3
```

Output:

```text
3 1
3 2
3 3
```

---

### So the complete output is:

```text
1 1
1 2
1 3

2 1
2 2
2 3

3 1
3 2
3 3
```

The important thing to remember is:

> **For every 1 repetition of the outer loop, the inner loop runs 3 times.**

Since the outer loop runs **3 times** and the inner loop runs **3 times**:

```text
3 × 3 = 9 total executions
```

# 🧑‍🎓 Think of Students and Subjects

Imagine you have **3 students**:

```text
Student 1
Student 2
Student 3
```

And each student has **3 subjects**:

```text
Math
English
Biology
```

You want to print every student's subjects.

```python
students = ["Ali", "Sara", "Ahmed"]
subjects = ["Math", "English", "Biology"]

for student in students:
    for subject in subjects:
        print(student, subject)
```

Output:

```text
Ali Math
Ali English
Ali Biology

Sara Math
Sara English
Sara Biology

Ahmed Math
Ahmed English
Ahmed Biology
```

Here:

```text
Outer loop → Students
Inner loop → Subjects
```

The inner loop goes through **all subjects for Ali**, then **all subjects for Sara**, then **all subjects for Ahmed**.
