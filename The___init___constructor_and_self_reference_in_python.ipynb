The __init__ constructor and self are two of the most important concepts in Python's Object-Oriented Programming (OOP).

Let's understand them step by step.



What is __init__?

The __init__ method is a special method (constructor) that is automatically called whenever you create an object from a class.

Its job is to initialize (give initial values to) the object's attributes.



```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age


s1 = Student("Ali", 20)
```

```python
#When you create an object:
s1 = Student("Ali", 20)
```

```python
#Python automatically does this:
Student.__init__(s1, "Ali", 20)
```

```python
#You never call __init__() yourself.
```

Why we use def?

def is used to define functions and methods

Inside a class, def defines methods (the actions an object perform)

```python
#Forexample
class Student:

    def __init__(self, name): #defines contructor; which sets up the object
        self.name = name

    def study(self):
        print(self.name, "is studying.") #defines a method that makes the student study

    def sleep(self):
        print(self.name, "is sleeping") #defines sleep method
```

```python
s1 = Student("Ali")
s1.study()
s1.sleep()
```

What is self?

self refers to the current object.

It lets each object store and access its own data.

Think of self as meaning:

"this particular object" or "Instance of object"

```python
#Example 1
class Student:

    def __init__(self, name):
        self.name = name

student1 = Student("Ali")
student2 = Student("Sara")

print(student1.name)
print(student2.name)
```

```python
#What happened?

#When you wrote
student1 = Student("Ali")
```

```python
#Python internally did:
self = student1
name = "Ali"
```

```python
#So inside __init__
self.name = name
```

```python
#became
student1.name = "Ali"
```

```python
#Then
student2 = Student("Sara")
```

```python
#became
student2.name = "Sara"
```

```python
#Each object gets its own name.
```

```python
#Methods also use self
class Student:

    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello", self.name)

s1 = Student("Ali")

s1.greet()
```

```python
#Python actually converts
s1.greet()
```

```python
#into
Student.greet(s1)
```

```python
#Another Example
class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

r = Rectangle(5, 4)

print(r.area())
```

```python
#Python internally does
Rectangle.area(r)
```

```python
#So self.length means r.length
```

```python
#Why is it called self?

#Actually, it doesn't have to be called self.

#This works:
class Test:

    def __init__(abc):
        print("Hello")
```

But by Python convention, we always use self because it makes code readable and is what other Python programmers expect.

```python
#Complete Example
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

emp1 = Employee("Ali", 50000)
emp2 = Employee("Fatima", 60000)

emp1.display()
print()

emp2.display()
```

```python
#Another example
class Cat:
    def __init__(self, name, colour):
        self.name = name
        self.colour = colour

    def meow(self):
        print(self.name, "says meow!")
```

```python
cat1 = Cat("Bella", "Simba")
print(cat1.colour)
cat1.meow()
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        self.balance -= amount
        print(f"Withdrew {amount}. New balance: {self.balance}")
```

```python
#Create an object
account1 = BankAccount("Ali", 50000)

print(account1.owner)
print(account1.balance)

account1.deposit(1000)
account1.withdraw(2000)
```

Key Points to Remember

__init__ is a constructor that runs automatically when an object is created.

self refers to the current object (instance).

self.attribute = value creates or updates an instance attribute.

Each object has its own separate copy of instance attributes.

Methods use self to access the object's data.

self is passed automatically when you call a method on an object.
