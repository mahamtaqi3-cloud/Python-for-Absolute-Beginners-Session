Modules & Packages in Python

1. Introduction: What is a Module?


Module: A single file containing Python code (functions, classes, variables) ending in .py.

Package: A directory of modules with a special __init__.py file (though in modern Python, this is often implicit).

Standard Library: Python "comes with batteries included." These are built-in modules you don't need to install.

They are ways to organize and reuse code.

```python
import math

# Constants
print(f"Pi: {math.pi}")
print(f"Euler's number: {math.e}")

# Functions
print(f"Ceiling of 4.2: {math.ceil(4.2)}")
print(f"Floor of 4.9: {math.floor(4.9)}")
print(f"Power: {math.pow(2, 3)}") # 2 to the power of 3
```

```python
#1. The math Module (Calculations)
#The math module is your best friend for anything involving geometry, trigonometry, or precision rounding.

#The "Circle Geometry" Example:
import math

radius = 5
# Calculating area: pi * r squared
area = math.pi * math.pow(radius, 2)

print(f"The area of a circle with radius {radius} is {area:.2f}")
print(f"Rounded down to the nearest integer: {math.floor(area)}")
```

```python
import random

# Random float between 0.0 and 1.0
print(f"Random float: {random.random()}")

# Random integer in a range
print(f"Dice roll: {random.randint(1, 6)}")

# Choosing from a list
players = ["Alice", "Bob", "Charlie", "Diana"]
print(f"Winner: {random.choice(players)}")

# Shuffling a list
random.shuffle(players)
print(f"Shuffled order: {players}")
```

The random Module (Logic & Probability)
This module is perfect for simulating games, selecting items, or generating test data.

The "Virtual Dice Roller" Example:

```python
import random

# Roll a 6-sided die
die_roll = random.randint(1, 6)

# Random choice from a list
options = ["Heads", "Tails"]
coin_flip = random.choice(options)

print(f"You rolled a: {die_roll}")
print(f"The coin landed on: {coin_flip}")
```

C. The datetime Module
Crucial for data analysis (the primary use case for Colab).

```python
import datetime as dt

# Get today's date
today = dt.date.today()
print(f"Today is: {today}")

# Get current time
now = dt.datetime.now()
print(f"Current time: {now}")

# Formatting dates
print(f"Formatted date: {now.strftime('%d/%m/%Y %H:%M')}")

# Time Deltas (math with dates)
tomorrow = today + dt.timedelta(days=1)
print(f"Tomorrow is: {tomorrow}")
```

3. The datetime Module (Time & Scheduling)
In Data Science (where Colab is king), you will constantly deal with timestamps.

The "Time Greeting" Example:

```python
import datetime

# Get the exact current moment
now = datetime.datetime.now()

# Extract just the hour
current_hour = now.hour

if current_hour < 12:
    print(f"Good morning! It is currently {now.strftime('%H:%M')}")
else:
    print(f"Good afternoon! It is currently {now.strftime('%H:%M')}")
```

4. The os Module (System Interaction)
This is crucial for Google Colab because it allows you to see where you are in the file system and interact with files.

The "File System Check" Example:

```python
import os

# Where am I? (In Colab, this is usually /content)
current_folder = os.getcwd()

print(f"Your code is running in: {current_folder}")

# What is in this folder?
print(f"Contents of the folder: {os.listdir(current_folder)}")
```
